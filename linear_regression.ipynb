First we create data simulating the salary of a worker based on their current years of experience

```python
from cProfile import label
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

rng = np.random.default_rng(seed=42) 
years = rng.uniform(0,12,size=30).round(1)
salary = 22_000 + 2_600 * years + rng.normal(0,3_000, size=30)
df = pd.DataFrame({"YearsExperience": years, "Salary": salary.round(-2)})
df.head()
```

```python
df.plot(
        kind='scatter',
        x='YearsExperience',
        y='Salary',
        title='Salary as a function of years of experience',
        grid=True,
        xlabel='Years of Experience',
        ylabel='Salary'
        );
```

Then we train a linear regresion model based on our input data to see how close is the model from our data and see the loss

```python
x = df['YearsExperience'].values
y = df['Salary'].values
model = LinearRegression().fit(x.reshape(-1, 1), y)
m = model.coef_[0] 
b = model.intercept_
y2 = m * x + b

x,y,y2
```

Now , once whe have de model trained, we may use it to predict which value is going to be for an specific input

```python
x_new = np.array([1.5, 2.5,4.5,5.5,6.5,10])
y2_new = m * x_new + b
y2_new = model.predict(x_new.reshape(-1,1))

plt.scatter(x, y, c='blue', label='Actual data')
plt.scatter(x_new, y2_new, c='red', label='Predictions')
plt.xlabel('Year of experience')
plt.ylabel('Salary')
plt.legend()
plt.grid(True)
plt.show()
```
